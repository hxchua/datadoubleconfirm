### This notebook shows how to perform the following data transformations to prepare the data for data visualizations or further analysis:   
- Get first row to be column headers
- Replace na with missing values
- Transpose columns to rows (change from wide to long format) 

### The dataset used is on Graduates From University First Degree Courses By Type Of Course And Sex in Singapore from [SingStat](https://tablebuilder.singstat.gov.sg/table/TS/M850511) and made available on datadoubleconfirm github repo [here](https://github.com/hxchua/datadoubleconfirm/blob/master/datasets/unigradsg.csv) in the format below.

```python
import numpy as np
```

```python
import pandas as pd
url = 'https://raw.githubusercontent.com/hxchua/datadoubleconfirm/refs/heads/master/datasets/unigradsg.csv'
df = pd.read_csv(url)
```

```python
df
```

```python
#get first row to be column headers
column_names = df.iloc[0].tolist()
df.columns = column_names
#remove first row
df = df[1:]
#replace na with missing values
df.replace("na", np.nan, inplace=True)
#transpose columns to rows
df = df.melt(id_vars=["Data Series"],
        var_name="Year",
        value_name="Value")
```
